```python
import os
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px
```

```python
heart_df = pd.read_csv("cardio_train.csv")
```

```python
heart_df.head(11)
```

```python
heart_df["age"] = (heart_df["age"] / 365).round().astype(int)
```

```python
heart_df["age"]
```

```python
heart_df
```

```python
heart_df.info()
```

```python
heart_df.isna().sum()
```

```python
correlation_matrix = heart_df.corr()
```

```python
plt.figure(figsize=(14,12))
sns.heatmap(correlation_matrix, cmap="Oranges", annot=True, fmt = ".2f", linewidths= .5)
plt.title("correlation matrix", fontsize = 16)
plt.savefig("images/heatmap.png")
```

```python
x = heart_df.iloc[:, 1:-1]
```

```python
x  
```

```python
y = heart_df["cardio"]
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x,y, test_size= 0.5, random_state=0)
x_train.shape
```

```python
from sklearn.preprocessing import MinMaxScaler 
```

```python
scaler = MinMaxScaler()
```

```python
scaler.fit(x)
```

```python
x_scaled =  scaler.transform(x) 
```

```python
x_scaled
```

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression(solver = "liblinear", penalty = "l2", max_iter=1000)
```

```python
trained_data = model.fit(x_train, y_train)
```

```python
y_pred = model.predict(x_test)
```

```python
y_pred
```

```python
from sklearn.metrics import accuracy_score, classification_report
```

```python
accuracy_score(y_test, y_pred)
```

```python
classification_report(y_test, y_pred)
```

```python
convo = [[40, 2, 157, 76, 141, 96, 2, 2, 0, 0, 1]]
```

```python
camel = trained_data.predict(convo)
```

```python
camel
```

```python
smaple_array = np.array(camel).reshape(1, -1)
```

```python
probabilties = model.predict_proba(convo)[0][1]
```

```python
percent = probabilties * 100
```

```python
print(f"probability of having {percent:.2f}%")
```

```python
import joblib
```

```python
heart_disease = {
    "model": model,
    "input_cols": x,
    "target_col": y,
    "train_input": x_train,
    "test_input": x_test,
    "train_target": y_train,
    "test_target": y_test,
    "accuracy": accuracy_score,
    "scaler": scaler
} 
```

```python
joblib.dump(heart_disease, "heart_disease.joblib")
```

```python
import matplotlib.pyplot as plt

colors = heart_df['alco'].map({0: 'darkorange', 1: 'sandybrown'})

plt.figure(figsize=(10, 6))
plt.scatter(heart_df['ap_lo'], heart_df['ap_hi'], c=colors, s=40, edgecolor='black', alpha=0.6)
plt.xlabel("API Low")
plt.ylabel("API High")
plt.title("API_hi vs API_lo by Alcohol Intake", fontsize=16, color="darkorange")
plt.grid(True, linestyle='--', alpha=0.4)
plt.savefig("images/api_scatter_plot.png")
plt.show()
```
